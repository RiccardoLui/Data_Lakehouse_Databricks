## Create gold fact_sales from silver tables

Select order number, product key, customer id, order date, ship date, due date, sales amount, quantity, price

```python
%sql
CREATE TABLE data_lakehouse_databricks.gold.gold_fact_sales 
AS SELECT 
    s.order_number,
    s.product_key,
    s.customer_id,
    s.order_date,
    s.ship_date,
    s.due_date,
    s.sales_amount,
    s.quantity,
    s.price

FROM data_lakehouse_databricks.silver.silver_sales_details s
```

```python
%sql
SELECT * FROM data_lakehouse_databricks.gold.gold_fact_sales LIMIT 10
```
