## Join silver tables to create gold materialized view

Select customer id (i), customer key (i,d,l), name (i), last name (i), dob (d), country (l), marital status (i), gender (d), creation date (i)

from info (i), demographics (d), location (l), join using customer key

```python
%sql
CREATE OR REPLACE TABLE data_lakehouse_databricks.gold.gold_dim_customers 
AS SELECT 
       i.customer_id,
       i.customer_key,
       i.firstname,
       i.lastname,
       i.marital_status,
       i.gender,

       l.country,

       d.birth_date,

       i.creation_date


FROM data_lakehouse_databricks.silver.silver_customer_info i

LEFT JOIN data_lakehouse_databricks.silver.silver_customer_demographics d
ON i.customer_key = d.customer_key

LEFT JOIN data_lakehouse_databricks.silver.silver_customer_location l 
ON i.customer_key = l.customer_key
```

```python
%sql
SELECT * FROM data_lakehouse_databricks.gold.gold_dim_customers limit 10
```
