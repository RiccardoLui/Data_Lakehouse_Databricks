# Gold customer dimension

Production notebook with the post-write preview query removed.


```python
spark.sql("""
CREATE OR REPLACE TABLE data_lakehouse_databricks.gold.gold_dim_customers
AS
SELECT
    i.customer_id,
    i.customer_key,
    i.firstname,
    i.lastname,
    i.marital_status,
    i.gender,
    l.country,
    d.birth_date,
    i.creation_date
FROM data_lakehouse_databricks.silver.silver_customer_info AS i
LEFT JOIN data_lakehouse_databricks.silver.silver_customer_demographics AS d
    ON i.customer_key = d.customer_key
LEFT JOIN data_lakehouse_databricks.silver.silver_customer_location AS l
    ON i.customer_key = l.customer_key
""")
```
