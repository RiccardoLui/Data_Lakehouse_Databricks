# Gold product dimension

Production notebook with the post-write preview query removed.


```python
spark.sql("""
CREATE OR REPLACE TABLE data_lakehouse_databricks.gold.gold_dim_product
AS
SELECT
    p.product_id,
    p.product_key,
    p.product_name,
    p.cost,
    p.product_line,
    p.start_date,
    p.end_date,
    p.category_id,
    c.category,
    c.subcategory,
    c.maintenance
FROM data_lakehouse_databricks.silver.silver_product_info AS p
LEFT JOIN data_lakehouse_databricks.silver.silver_product_category AS c
    ON p.category_id = c.category_id
""")
```
