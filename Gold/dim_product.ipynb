## Create gold table from silver layer

select product id (p), product key (p), product name (p), cost (p), product line (p), start date (p), end date (p), category_id (p,c), category (c), subcategory (c), maintenance (c)

```python
%sql
CREATE OR REPLACE TABLE data_lakehouse_databricks.gold.gold_dim_product
AS SELECT
    p.product_id,
    p.product_key,
    p.product_name,
    p.cost,
    p.product_line,
    p.start_date,
    p.end_date,
    p.category_id,

    c.category,
    c.subcategory,
    c.maintenance

FROM data_lakehouse_databricks.silver.silver_product_info p
LEFT JOIN data_lakehouse_databricks.silver.silver_product_category c
ON p.category_id = c.category_id
```

```python
%sql
SELECT * FROM data_lakehouse_databricks.gold.gold_dim_product LIMIT 10
```
